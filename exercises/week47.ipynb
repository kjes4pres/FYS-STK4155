### Question 9:

Consider a fully-connected network with layer sizes $n_0$ (the input
layer) ,$n_1$ (first hidden layer), $\dots, n_L$, where $n_L$ is the
output layer. Derive a general formula for the total number of
trainable parameters (weights + biases).

__Answer__

Each layer has a weight matrix and a bias vector, which align with the size of the layer. So to get to the number of trainable parameters, we have to sum them over the number of layers. Every neuron in layer $l$ connects to all neurons in the previous layer $l-1$, in addition each neuron in layer $l$ has a corresponding bias. So for layer $l$, the weight matrix is of size $(n_l, n_{l-1}) and the bias $(n_l, 1)$.

So in total: 

$$\text{Number of parameters} = \sum_{l=1}^{L} (n_l\cdot n_{l-1} + n_l) $$
